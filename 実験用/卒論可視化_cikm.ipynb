```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib
import matplotlib.ticker as mticker


def load_filter_and_plot(
    *,
    path,
    csv_name,
    dataset,
    x_col,
    defaults,
    y_col="score_mean",
    for_slide=False,
    only_ticks=True,
    y_ticks=None,
    error_mode=None,               # None | "sd" | "ci95"
    error_col="score_stdev",      # 誤差元列（標準偏差）
    ci_n_col="seed_range",        # ci95 用 n 列
    capsize=3,
):
    # =====================
    # load
    # =====================
    csv_path = os.path.join(path, csv_name)
    df = pd.read_csv(csv_path)

    # 数値化（FutureWarning 回避）
    for col in df.columns:
        try:
            df[col] = pd.to_numeric(df[col])
        except (ValueError, TypeError):
            pass

    if "G_type" not in df.columns:
        raise ValueError("CSV must contain 'G_type' column.")
    df["G_type"] = df["G_type"].astype(str)

    # =====================
    # DEFAULTS fillna
    # =====================
    for key, val in defaults.items():
        if key in df.columns and val is not None:
            df[key] = df[key].fillna(val)

    # =====================
    # DEFAULTS filtering
    # =====================
    for key, val in defaults.items():
        if key == x_col:
            continue
        if key not in df.columns or val is None:
            continue
        df = df[df[key] == val]

    if df.empty:
        raise ValueError("No data left after DEFAULTS filtering.")

    # =====================
    # G_type 名称統一
    # =====================
    df.loc[df["G_type"] == "nonlinear_new", "G_type"] = "nonlinear"
    df.loc[df["G_type"] == "laplacian_nonlinear_new", "G_type"] = "laplacian_nonlinear"

    # =====================
    # zerosum / regularization → G_type
    # =====================
    reg_key = (
        df["regularization"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
    ) if "regularization" in df.columns else pd.Series("", index=df.index)

    is_tg = reg_key.isin(["target-graph", "target_graph"])
    is_penal = reg_key.isin(["penal-target-graph", "penal_target_graph"])

    is_zero = (df["zerosum"] == True) if "zerosum" in df.columns else pd.Series(False, index=df.index)

    # nonlinear
    df.loc[(df["G_type"] == "nonlinear") & is_zero, "G_type"] = "nonlinear_zero"

    # laplacian_nonlinear
    df.loc[df["G_type"] == "laplacian_nonlinear_new", "G_type"] = "laplacian_nonlinear"

    # penal-target-graph
    df.loc[(df["G_type"] == "laplacian_nonlinear") & ~is_zero & is_penal, "G_type"] = "laplacian_nonlinear_penal"
    df.loc[(df["G_type"] == "laplacian_nonlinear") & is_zero & is_penal, "G_type"] = "laplacian_nonlinear_zero_penal"

    # target-graph
    df.loc[(df["G_type"] == "laplacian_nonlinear") & ~is_zero & is_tg, "G_type"] = "laplacian_nonlinear_tg"
    df.loc[(df["G_type"] == "laplacian_nonlinear") & is_zero & is_tg, "G_type"] = "laplacian_nonlinear_zero_tg"

    # plain graph/others
    df.loc[(df["G_type"] == "laplacian_nonlinear") & is_zero & ~is_tg & ~is_penal, "G_type"] = "laplacian_nonlinear_zero"

# =====================
    # axis style
    # =====================
    x_style = X_AXIS_STYLE.get(x_col, {})
    xlabel = x_style.get("xlabel", x_col)
    xscale = x_style.get("xscale", "linear")
    base_title = x_style.get("title", None)

    y_style = Y_AXIS_STYLE.get(y_col, {})
    ylabel = y_style.get("ylabel", y_col)

    ticks = x_style.get("ticks", None)
    tick_labels = x_style.get("labels", None)

    if only_ticks and ticks is None:
        raise ValueError(f"only_ticks=True but no ticks defined for x_col='{x_col}'")

    title = f"{dataset}：{base_title}" if base_title else dataset

    # =====================
    # style switch (paper / slide)
    # =====================
    if for_slide:
        FIGSIZE = (10, 7)
        FONT_SIZE = {"title": 25, "label": 25, "tick": 20, "legend": 20}
        LINE_WIDTH = 5.0
        MARKER_SIZE = 15
        GRID_ALPHA = 0.35
        line_alpha = 0.85
        lw = LINE_WIDTH * 0.8
    else:
        FIGSIZE = (9, 6)
        FONT_SIZE = {"title": 20, "label": 25, "tick": 20, "legend": 18}
        LINE_WIDTH = 3.0
        MARKER_SIZE = 10
        GRID_ALPHA = 0.5
        line_alpha = 0.9
        lw = LINE_WIDTH

    # =====================
    # plot
    # =====================
    plt.figure(figsize=FIGSIZE)

    handles = []
    legend_labels = []

    if for_slide:
        allowed_gtypes = x_style.get("G_type_for_slide", None)
        if allowed_gtypes is None:
            allowed_gtypes = x_style.get("G_type", None)
    else:
        allowed_gtypes = x_style.get("G_type", None)

    if allowed_gtypes is not None:
        allowed_gtypes = set(allowed_gtypes)

    if only_ticks:
        ticks_rounded = [round(float(t), 6) for t in ticks]

    use_err = error_mode in {"sd", "ci95"}
    if use_err and error_col not in df.columns:
        raise ValueError(f"error_col '{error_col}' is missing in CSV.")
    if error_mode == "ci95" and ci_n_col not in df.columns:
        raise ValueError(f"ci95 requires ci_n_col='{ci_n_col}' in CSV.")

    for g, st in STYLE.items():
        if allowed_gtypes is not None and g not in allowed_gtypes:
            continue
        if g not in df["G_type"].unique():
            continue

        gdf = (
            df[df["G_type"] == g]
            .dropna(subset=[x_col, y_col])
            .sort_values(x_col)
        )

        if only_ticks:
            gdf = gdf[gdf[x_col].astype(float).round(6).isin(ticks_rounded)]

        if gdf.empty:
            continue

        if use_err:
            agg_cols = {y_col: "mean", error_col: "mean"}
            if error_mode == "ci95":
                agg_cols[ci_n_col] = "mean"

            gplot = gdf.groupby(x_col, as_index=False).agg(agg_cols).sort_values(x_col)
            y = gplot[y_col].astype(float).to_numpy()
            sd = gplot[error_col].astype(float).to_numpy()

            if error_mode == "sd":
                yerr = sd
            else:
                n = gplot[ci_n_col].astype(float).to_numpy()
                n = np.where(np.isfinite(n) & (n > 0), n, 1.0)
                yerr = 1.96 * (sd / np.sqrt(n))

            eb = plt.errorbar(
                gplot[x_col],
                y,
                yerr=yerr,
                color=st["color"],
                linestyle=st["linestyle"],
                marker=st["marker"],
                linewidth=lw,
                markersize=MARKER_SIZE,
                alpha=line_alpha,
                capsize=capsize,
                elinewidth=max(1.0, lw * 0.45),
            )
            line = eb.lines[0]
        else:
            gplot = gdf.groupby(x_col, as_index=False)[y_col].mean().sort_values(x_col)
            line, = plt.plot(
                gplot[x_col],
                gplot[y_col],
                color=st["color"],
                linestyle=st["linestyle"],
                marker=st["marker"],
                linewidth=lw,
                markersize=MARKER_SIZE,
                alpha=line_alpha,
            )

        handles.append(line)
        legend_labels.append(st["name"])

    plt.xscale(xscale)
    ax = plt.gca()

    # =====================
    # ticks / labels 反映（優先）
    # =====================
    if ticks is not None:
        if xscale == "log":
            ticks = [t for t in ticks if t > 0]

        ax.xaxis.set_major_locator(mticker.FixedLocator(ticks))

        if tick_labels is None:
            tick_labels = [f"{t:g}" for t in ticks]

        ax.xaxis.set_major_formatter(mticker.FixedFormatter(tick_labels))
        ax.xaxis.set_minor_locator(mticker.NullLocator())
    else:
        x_values = (
            df.dropna(subset=[x_col])
            .sort_values(x_col)[x_col]
            .astype(float)
            .round(6)
            .unique()
        )
        if xscale == "log":
            x_values = [x for x in x_values if x > 0]

        ax.xaxis.set_major_locator(mticker.FixedLocator(x_values))
        ax.xaxis.set_major_formatter(mticker.FixedFormatter([f"{x:g}" for x in x_values]))
        ax.xaxis.set_minor_locator(mticker.NullLocator())

    plt.xlabel(xlabel, fontsize=FONT_SIZE["label"])
    if error_mode == "sd":
        ylabel = f"{ylabel}（±SD）"
    elif error_mode == "ci95":
        ylabel = f"{ylabel}（±95%CI）"
    plt.ylabel(ylabel, fontsize=FONT_SIZE["label"])
    plt.title(title, fontsize=FONT_SIZE["title"])

    # =====================
    # y-axis control
    # =====================
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter('%.2f'))
    ax.tick_params(axis="both", labelsize=FONT_SIZE["tick"])

    if y_ticks is not None:
        if len(y_ticks) != 2:
            raise ValueError("y_ticks must be [ymin, ymax]")

        ymin, ymax = y_ticks
        ax.set_ylim(ymin, ymax)
        ax.yaxis.set_major_locator(mticker.MaxNLocator(nbins=5))

    # legend（for_slide は無し）
    if (not for_slide) and handles:
        plt.legend(
            handles,
            legend_labels,
            loc=LEGEND_STYLE.get(x_col, "upper left"),
            fontsize=FONT_SIZE["legend"],
            frameon=True,
        )

    plt.grid(True, which="both", ls="--", alpha=GRID_ALPHA)
    plt.tight_layout()
    plt.show()


```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import betaln

# パラメータ
betas = [1e-2, 1e-1, 1, 10, 100]
x = np.linspace(1e-6, 1 - 1e-6, 8000)

# Beta(a,a) の pdf（数値安定化のため log で計算）
def beta_pdf(x, beta):
    return np.exp((beta - 1) * np.log(x) +
                  (beta - 1) * np.log(1 - x) -
                  betaln(beta, beta))

# 描画
plt.figure(figsize=(9, 5.5))
for b in betas:
    plt.plot(x, beta_pdf(x, b), label=f"Beta({b:g}, {b:g})")

# 日本語ラベル
plt.title("対称 Beta 分布（線形スケール）", fontsize=20)
plt.xlabel("ラベル1の比率 $x$（2ラベルの場合）", fontsize=25)
plt.ylabel("確率密度関数（pdf）", fontsize=25)

plt.ylim(0, 15)
plt.legend(title="分布", fontsize=20, title_fontsize=20)
plt.grid(True, linewidth=0.4)
plt.xticks(fontsize=22)
plt.yticks(fontsize=22)

plt.tight_layout()
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

# パラメータ
betas = [1e-2, 1e-1, 1, 10, 100]
L = 3
n_samples = 5000

# 正三角形（3ラベル単体）の頂点
V = np.array([
    [0.0, 0.0],                 # ラベル1
    [1.0, 0.0],                 # ラベル2
    [0.5, np.sqrt(3) / 2],      # ラベル3
])

def simplex_projection(p):
    """3次元Dirichletサンプルを2次元単体へ射影"""
    return p @ V

fig, axes = plt.subplots(1, len(betas), figsize=(20, 6),
                         sharex=True, sharey=True)

for ax, beta in zip(axes, betas):
    samples = np.random.dirichlet([beta] * L, size=n_samples)
    xy = simplex_projection(samples)

    ax.scatter(xy[:, 0], xy[:, 1], s=1, alpha=0.9)
    ax.set_title(rf"$\beta={beta:g}$", fontsize=50)
    ax.set_aspect("equal")
    ax.axis("off")

    # 単体（正三角形）の枠線
    tri = np.vstack([V, V[0]])
    ax.plot(tri[:, 0], tri[:, 1], color="black", linewidth=0.4)

fig.suptitle("3ラベルにおける Dirichlet($\\beta,\\beta,\\beta$) 分布の可視化",
             fontsize=50)
plt.tight_layout()
plt.show()

```

```python
def load_filter_and_plot_diff_from_base(
    path,
    csv_name='output.csv',
    dataset='mnist',
    x_col='num_institution',
    base_G_type='nonlinear_zero',
    defaults=None,
    y_col='???',
    seed_col=None,
    g_types=None,
    style_map=None,
    title=None,
):
    """
    base_G_type を基準に、同一 seed の y_col 差分 (target - base) を計算して
    x_col ごとの平均±標準偏差をプロットする。
    """
    import os
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt

    defaults = defaults or {}

    csv_path = os.path.join(path, csv_name)
    df = pd.read_csv(csv_path, encoding='utf-8-sig')

    if 'dataset' in df.columns:
        df = df[df['dataset'] == dataset]

    for key, val in defaults.items():
        if key in df.columns:
            df = df[df[key] == val]

    if df.empty:
        raise ValueError('No data left after dataset/default filtering.')

    # 既存 load_filter_and_plot と同様の G_type 正規化
    if 'G_type' not in df.columns:
        raise ValueError("CSV must contain 'G_type' column.")
    df['G_type'] = df['G_type'].astype(str)
    df.loc[df['G_type'] == 'nonlinear_new', 'G_type'] = 'nonlinear'
    df.loc[df['G_type'] == 'laplacian_nonlinear_new', 'G_type'] = 'laplacian_nonlinear'

    if 'zerosum' in df.columns:
        z = (
            df['zerosum']
            .astype(str)
            .str.lower()
            .map({'true': True, 'false': False})
            .fillna(df['zerosum'])
        )
        z = z.astype(bool)
    else:
        z = pd.Series(False, index=df.index)

    reg = (
        df['regularization'].astype(str).str.lower()
        if 'regularization' in df.columns
        else pd.Series('identity', index=df.index)
    )

    is_tg = reg.eq('target-graph')
    is_penal = reg.eq('target')

    df.loc[(df['G_type'] == 'nonlinear') & z, 'G_type'] = 'nonlinear_zero'
    df.loc[(df['G_type'] == 'laplacian_nonlinear') & z & ~is_tg & ~is_penal, 'G_type'] = 'laplacian_nonlinear_zero'
    df.loc[(df['G_type'] == 'laplacian_nonlinear') & ~z & is_tg, 'G_type'] = 'laplacian_nonlinear_tg'
    df.loc[(df['G_type'] == 'laplacian_nonlinear') & z & is_tg, 'G_type'] = 'laplacian_nonlinear_zero_tg'
    df.loc[(df['G_type'] == 'laplacian_nonlinear') & ~z & is_penal, 'G_type'] = 'laplacian_nonlinear_penal'
    df.loc[(df['G_type'] == 'laplacian_nonlinear') & z & is_penal, 'G_type'] = 'laplacian_nonlinear_zero_penal'

    if y_col not in df.columns:
        raise ValueError(f"'{y_col}' column not found in CSV.")
    if x_col not in df.columns:
        raise ValueError(f"'{x_col}' column not found in CSV.")

    if seed_col is None:
        for c in ['seeds', 'seed_values', 'seed', 'seed_range']:
            if c in df.columns:
                seed_col = c
                break
    if seed_col is None:
        raise ValueError('Seed column not found. Please pass seed_col explicitly.')

    pair_cols = [x_col, seed_col]

    # 重複を避けるため、同一 pair 内は平均して1点化
    reduced = (
        df[[*pair_cols, 'G_type', y_col]]
        .dropna(subset=[x_col, seed_col, y_col])
        .groupby([*pair_cols, 'G_type'], as_index=False)[y_col]
        .mean()
    )

    base_df = (
        reduced[reduced['G_type'] == base_G_type][pair_cols + [y_col]]
        .rename(columns={y_col: 'base_val'})
    )
    if base_df.empty:
        raise ValueError(f"base_G_type='{base_G_type}' has no rows after filtering.")

    if g_types is None:
        g_types = [g for g in sorted(reduced['G_type'].unique()) if g != base_G_type]

    plt.figure(figsize=(8, 5))
    plotted = False

    for g in g_types:
        gdf = (
            reduced[reduced['G_type'] == g][pair_cols + [y_col]]
            .rename(columns={y_col: 'target_val'})
        )
        if gdf.empty:
            continue

        merged = pd.merge(gdf, base_df, on=pair_cols, how='inner')
        if merged.empty:
            continue

        merged['diff'] = merged['target_val'] - merged['base_val']

        stats = (
            merged.groupby(x_col, as_index=False)['diff']
            .agg(mean='mean', std='std', n='count')
            .sort_values(x_col)
        )
        stats['std'] = stats['std'].fillna(0.0)

        if style_map and g in style_map:
            st = style_map[g]
            label = st.get('name', g)
            color = st.get('color', None)
            ls = st.get('linestyle', '-')
            mk = st.get('marker', 'o')
        else:
            label, color, ls, mk = g, None, '-', 'o'

        plt.errorbar(
            stats[x_col],
            stats['mean'],
            yerr=stats['std'],
            label=label,
            color=color,
            linestyle=ls,
            marker=mk,
            capsize=4,
        )
        plotted = True

    plt.axhline(0.0, color='gray', linestyle='--', linewidth=1)
    plt.xlabel(x_col)
    plt.ylabel(f"{y_col} difference (target - {base_G_type})")
    plt.title(title or f"Difference from base_G_type={base_G_type} ({dataset})")
    plt.grid(True, alpha=0.3)
    if plotted:
        plt.legend(loc='best')
    plt.tight_layout()
    plt.show()

```
