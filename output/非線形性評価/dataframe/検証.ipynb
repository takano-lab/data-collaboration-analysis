```python
import pandas as pd
df1 = pd.read_csv("test_df_letter_recognition_kernel_pca_self_tuning_GEP_1.csv")
df2 = pd.read_csv("test_df_letter_recognition_kernel_pca_self_tuning_nonlinear_1.csv")
# check exact equality
if df1.equals(df2):
    print("df1 and df2 are exactly identical")
else:
    print("df1 and df2 are NOT identical")
    print("Shapes:", df1.shape, df2.shape)
    print("Column order equal:", list(df1.columns) == list(df2.columns))
    print("Dtypes equal:", df1.dtypes.equals(df2.dtypes))
    # show differences (pandas >= 1.1.0)
    try:
        diff = df1.compare(df2)
        print("First differing entries (up to 10 rows):")
        display(diff.head(10))
    except Exception:
        # fallback: show rows with any difference
        neq_rows = (df1 != df2) & ~(df1.isna() & df2.isna())
        rows_with_diff = df1.index[neq_rows.any(axis=1)].tolist()
        print("Number of rows with any difference:", len(rows_with_diff))
        print("First differing row indices (up to 20):", rows_with_diff[:20])
```
